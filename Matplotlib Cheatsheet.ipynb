# MATPLOTLIB COMPLETE CHEATSHEET

```python
# Matplotlib = Python's primary plotting/visualization library
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

# BASIC PLOTTING CONCEPTS

```python
# Create sample data
x = np.arange(1, 6)
y = x ** 2

print("x:", x)
print("y:", y)
```

# BASIC LINE PLOT

```python
# plot() = draws a line graph
plt.plot(x, y)

# Title of the plot
plt.title("Simple Line Plot")

# X-axis label
plt.xlabel("X values")

# Y-axis label
plt.ylabel("Y values")

print("Line plot created: y = x^2")

plt.show()
```

# MULTIPLE LINES IN ONE PLOT

```python
y2 = x ** 3

# Plot multiple lines
plt.plot(x, y, label="x squared")
plt.plot(x, y2, label="x cubed")

# Show legend
plt.legend()
# X-axis label
plt.xlabel("X values")

# Y-axis label
plt.ylabel("Y values")

print("Two lines plotted: x^2 and x^3")

plt.show()
```

# LINE STYLES, COLORS & MARKERS

```python
# linestyle, color, marker customization
plt.plot(x, y, linestyle='--', color='red', marker='o')

print("Line style: dashed, color: red, marker: circle")

plt.show()
```

# SCATTER PLOT

```python
# scatter() = shows individual data points
plt.scatter(x, y, color='green',marker='o')

plt.title("Scatter Plot")
print("Scatter plot created")

plt.show()
```

# BAR CHART

```python
# bar() = vertical bar chart
plt.figure(dpi = 50)
plt.bar(x, y, label = "squared", color = 'red')
plt.legend()
plt.title("Bar Chart")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
print("Bar chart showing x vs y")

plt.show()
```

# HORIZONTAL BAR CHART

```python
# barh() = horizontal bar chart
plt.barh(x, y, label = "Squared",color = 'green')
plt.legend()
print("Horizontal bar chart created")

plt.show()
```

# HISTOGRAM

```python
data = np.random.randn(100)

# hist() = frequency distribution
plt.hist(data, bins=10)

print("Histogram with 10 bins")

plt.show()
```

# PIE CHART

```python
sizes = [40, 30, 20, 10]
labels = ['A', 'B', 'C', 'D']

# pie() = circular percentage chart
plt.pie(sizes, labels=labels, autopct='%1.1f%%')

print("Pie chart showing percentage distribution")

plt.show()
```

# SUBPLOTS (MULTIPLE GRAPHS)

```python
# subplot(rows, cols, index)
plt.subplot(1, 2, 1)
plt.plot(x, y)
plt.title("x^2")

plt.subplot(1, 2, 2)
plt.plot(x, y2)
plt.title("x^3")

print("Two subplots created in one figure")

plt.show()
```

# FIGURE & AXES OBJECTS (OO INTERFACE)

```python
# Object-oriented plotting (recommended for complex plots)
fig, ax = plt.subplots()

ax.plot(x, y)
ax.set_title("OO Style Plot")
ax.set_xlabel("X")
ax.set_ylabel("Y")

print("Plot created using Figure and Axes objects")

plt.show()
```

# FIGURE SIZE & DPI

```python
plt.figure(figsize=(6,4), dpi=100)
plt.plot(x, y)

print("Custom figure size and resolution set")

plt.show()
```

# GRID & STYLE

```python
# grid() = background grid lines
plt.plot(x, y)
plt.grid(True)

print("Grid enabled on plot")

plt.show()
```

# AXIS LIMITS & SCALE

```python
plt.plot(x, y)

# Set axis limits
plt.xlim(0, 10)
plt.ylim(0, 60)

print("Axis limits adjusted")

plt.show()
```

# LOG SCALE

```python
# y-axis logarithmic scale
plt.plot(x, y)
plt.yscale('log')
plt.xscale('log')

print("Y-axis set to logarithmic scale")

plt.show()
```

# ANNOTATIONS & TEXT

```python
plt.plot(x, y)

# annotate() = label a point
plt.annotate("Peak", xy=(5,25), xytext=(3,20),
             arrowprops=dict(arrowstyle="->"))

print("Annotation added to plot")

plt.show()
```

# LEGEND

```python
plt.plot(x, y, label="x^2")
plt.plot(x, y2, label="x^3")
plt.legend()

print("Legend added")

plt.show()
```

# TICKS & LABEL ROTATION

```python
plt.plot(x, y)

# Rotate x-axis labels
plt.xticks(rotation=45)

print("X-axis labels rotated")

plt.show()
```

# SAVE FIGURE

```python
plt.plot(x, y)

# Save plot to file
plt.savefig("my_plot.png")

print("Plot saved as my_plot.png")

plt.show()
```

# CLEAR, CLOSE & RESET

```python
plt.plot(x, y)
plt.clf()   # clear current figure
plt.close() # close figure window

print("Figure cleared and closed")
```

# FILL BETWEEN (AREA PLOT)

```python
plt.plot(x, y)
plt.fill_between(x, y, alpha=0.8)

print("Area under curve filled")

plt.show()
```

# STACK PLOT

```python
y1 = x
y2 = x + 2
y3 = x + 4

plt.stackplot(x, y1, y2, y3)

print("Stack plot created")

plt.show()
```

# STEP PLOT

```python
# step() = step-wise plot
plt.step(x, y)

print("Step plot created")

plt.show()
```

# ERROR BAR PLOT

```python
error = [1, 2, 1, 2, 1]

plt.errorbar(x, y, yerr=error, fmt='o')

print("Error bar plot created")

plt.show()
```

# STEM PLOT

```python
# stem() = discrete signal plot
plt.stem(x, y)

print("Stem plot created")

plt.show()
```

# BOX PLOT

```python
data = [np.random.randn(100), np.random.randn(100)]

plt.boxplot(data)

print("Box plot showing data distribution")

plt.show()
```

# VIOLIN PLOT

```python
plt.violinplot(data)

print("Violin plot created")

plt.show()
```

# IMAGES & HEATMAPS

```python
img = np.random.rand(10, 10)

# imshow() = display image / matrix
plt.imshow(img, cmap='viridis')
plt.colorbar()

print("Heatmap displayed using imshow")

plt.show()
```

# COLORMAPS

```python
plt.imshow(img, cmap='plasma')

print("Using plasma colormap")

plt.show()
```

# 3D PLOTTING

# STYLES & THEMES

```python
# Apply predefined style
plt.style.use('seaborn-v0_8')

plt.plot(x, y)

print("ggplot style applied")

plt.show()
```

# INTERACTIVE MODE

```python
plt.ion()  # turn on interactive mode
plt.plot(x, y)

print("Interactive mode ON")

plt.ioff() # turn off interactive mode
```

# TIGHT LAYOUT


```python
plt.plot(x, y)
plt.tight_layout()

print("Layout adjusted automatically")

plt.show()
```

# TWIN AXES

```python
fig, ax1 = plt.subplots()
ax2 = ax1.twinx()

ax1.plot(x, y, color='blue')
ax2.plot(x, y2, color='red')

print("Two Y-axes on same plot")

plt.show()
```

# EVENT HANDLING (ADVANCED)

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(10)
y = x**2

def onclick(event):
    if event.inaxes:
        print("Mouse clicked at:", event.xdata, event.ydata)

fig, ax = plt.subplots()
fig.canvas.mpl_connect('button_press_event', onclick)

ax.plot(x, y)
print("Click on plot to see coordinates")
plt.show()
```
