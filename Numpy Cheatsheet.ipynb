# Array Creation - From list / tuple

```python
a = np.array([1, 2, 3])
b = np.array((4, 5, 6))
print(a)
print(b)
```

# Zeros, Ones, Empty

```python
c = np.zeros(5)
d = np.ones((2,3))
e = np.empty(4)
print(c)
print(d)
print(e)
```

# Range-based

```python
a = np.arange(0, 10, 2)
b = np.linspace(0, 1, 5)
print(a)
print(b)
```

# Identity / Diagonal

```python
a = np.eye(3)
b = np.identity(3)
c = np.diag([1,2,3])
print(a)
print(b)
print(c)
```

# Array Attributes

```python
a = np.array([[2,3],
     [4,5]])
print(a.ndim)     # dimensions
print(a.shape)     # rows, columns
print(a.size)      # total elements
print(a.dtype)     # data type
print(a.itemsize)  # bytes per element
```

# Data Types

```python
a = np.array([1,2,3], dtype=float)
b = np.array([1,2,3], dtype=int)
c = np.array([1,0,1], dtype=bool)
print(a)
print(b)
print(c)
# Type conversion:
d = c.astype(float)
print(d)
```

# Indexing & Slicing

```python
# 1D Indexing
a = np.array([10,20,30,40,50])

print(a[0])
print(a[-1])
print(a[1:4])
print(a[:3])
print(a[::2])
```

```python
# 2D Indexing
b = np.array([[1,2],[3,4],[5,6]])
print(b[0,1])
print(b[:,0])
print(b[1,:])
```

# Array Reshaping

```python
a = np.array([10,20,30,40,50,60])
print(a.reshape(2,3)) # Changes the shape of the array
print(a.flatten()) # Converts array into 1D. Always returns a copy (new memory)
print(a.ravel()) # Converts array into 1D. Returns a view if possible (not a copy)
b = np.array([[1,2],
            [3,4],
            [5,6]])
print(b.transpose()) # Swaps axes (rows ↔ columns)
print(b.T) # Shortcut (property) for transpose()
```

# Mathematical Operations

```python
# Arithmetic
x = np.array([[1,2],
            [3,4]])
y = np.array([[5,6],
            [7,8]])
print(np.add(x, y))        # x + y
print(np.subtract(x, y))    # x - y
print(np.multiply(x, y))     # x * y
print(np.divide(x, y))       # x / y
print(np.power(x, y))        # x ** y
print(np.mod(x, y))          # remainder
print(np.remainder(x, y))
print(np.floor_divide(x, y))
```

```python
# Rounding & Approximation
x = [1,2,3.3,4.5,5.5,6.5,7.7,8]
print(np.round(x)) # round to nearest even
print(np.around(x)) # round is just an alias for around
print(np.floor(x)) 
print(np.ceil(x))
print(np.trunc(x))
print(np.fix(x))
```

```python
# Exponential & Logarithmic
np.exp(x)
np.exp2(x)
np.expm1(x)

np.log(x)
np.log10(x)
np.log2(x)
np.log1p(x)
```

```python
# POWER & ROOT OPERATIONS
x = np.array([1,2,3,4])
y = np.array([5,6,7,8])
np.sqrt(x)
np.cbrt(x)
np.square(x)
np.power(x, 2)
np.hypot(x, y)
```

```python
# TRIGONOMETRIC FUNCTIONS
np.sin(a)
np.cos(a)
np.tan(a)
```

```python
# Inverse
x = np.array([1])
y = np.array([5])
np.arcsin(x)
np.arccos(x)
np.arctan(x)
np.arctan2(x, y)
```

```python
# Hyperbolic
a = np.array([180])
b = np.array([60])
np.sinh(a)
np.cosh(a)
np.tanh(a)
np.arcsinh(a)
np.arccosh(a)
np.arctanh(a)
```

```python
# DEGREE ↔ RADIAN CONVERSION
np.deg2rad(a)
np.rad2deg(a)
np.degrees(a)
np.radians(a)
```
